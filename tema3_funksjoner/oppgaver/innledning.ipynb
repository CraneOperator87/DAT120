```python
# eget oppgave 1 

def bokbestilling ():
    tittel = input("Tittel på boken: ")
    forfatter = input("Navnet på forfatter: ")
    eksemplar = int(input("Hvor mange?: "))

    brukerinput = input("Har du prisen totalt, eller per bok?").lower()
    pris = float(input("Pris: "))

    enhets_pris, total_pris = kost(
        pris, eksemplar, brukerinput
    )

    return formatter(
        tittel, forfatter, eksemplar, enhets_pris, total_pris
    )
    

def kost(pris, eksemplar, pristype):  
    if pristype == "totalt":
        totalt_pris = pris
        enhets_pris = pris / eksemplar
    else:
        enhets_pris = pris
        totalt_pris = pris * eksemplar

    return enhets_pris, totalt_pris
      
def formatter(tittel, forfatter, eksemplarer, enhets_pris, totalt_pris):
        
        
        return f"""bokbestilling:
    Tittel: {tittel}
    Forfatter: {forfatter}
    Antall: {eksemplarer}
    Pris per bok: {enhets_pris:.2f}
    Total sum: {totalt_pris:.2f}"""

print(bokbestilling())
```

```python
def utleie():
    godkjent = ["e","d","b","elektrisk","bensin","diesel","strøm"]
    while True:
        try:
            merke = input(
                 "Hvilke merke er bilen din?: "
            ).lower()
            modell = input(
                 f"Hvilke modell er {merke}-en din?: "
            )

            while True:
                try:
                    aarsmodell = int(input(
                        "Fra hvilket år er bilen din?: "
                    ))
                    if aarsmodell > 1885:
                        break
                    print(
                        "Skriv inn et gyldig årsmodell."
                        )
                except ValueError:
                    print (
                        "Ugyldig Syntax! Prøv på nytt."
                        )
                    
            farge = input(
                 f"Hva er fargen på din {modell}?: "
            )
            while True:
                drivstoff = input(
                    "[B / Bensin, D/Diesel, E/Elektrisk/Strøm]\nHvilken drivstoff bruker bilen din?: "
                ).lower()
                if drivstoff in godkjent:
                    while True:
                        try:
                            kapasitet = int(input(
                                "Hvor stor kapasitet har bilen din? (L/kW):"
                            ))
                            rekkevidde = float(input(
                                "Hva er rekkevidden til bilen din med full kapasitet?: "
                            ))
                            break
                        except ValueError:
                            print(
                                "Ugyldig Syntax! Svar med et heltall!"
                            ) 
                else:
                    print(
                        "Vennligst svar B,D eller E. Prøv på nytt."
                    )
                    continue
                break
            while True:
                try:
                    kjort = int(input(
                    "Hvor mange kilometer kjørte du bilen under leien?: "
                    ))
                    if kjort > 1:
                        break
                    print(
                        "Vennligst skriv en gyldig KM-avstand."
                    )
                except ValueError:
                    print (
                        "Ugyldig Syntax! Prøv på nytt."
                    )

        except ValueError:
            print (
                "Ugyldig Syntax! Prøv på nytt."
                )

        return merke, modell, aarsmodell, farge, drivstoff, kapasitet, rekkevidde, kjort

    

def kost(drivstoff, kjort):
    if drivstoff in ("e", "elektrisk"):
        pris_per_km = 5.59 # kr,-
    elif drivstoff in ("b", "bensin"):
        pris_per_km = 23.78 # kr,-
    elif drivstoff in ("d", "diesel"):
        pris_per_km = 20.00 # kr,-
    else:
        return "Ugyldig drivstoff"

    return pris_per_km * kjort

def main():
    merke, modell, aarsmodell, farge, drivstoff, kapasitet, rekkevidde, kjort = utleie()
    total_pris = kost(drivstoff, kjort)

    print(
        f"Du har leid en {aarsmodell} {merke} {modell} i fargen {farge}.\n"
        f"Drivstoff: {drivstoff}\n"
        f"Kapasitet: {kapasitet} L/kW\n"
        f"Rekkevidde: {rekkevidde} km\n"
        f"Du har kjørt {kjort} km under leien.\n"
        f"Total pris for leie: {total_pris:.2f} kr."
    )

if __name__ == "__main__":
    main()

```
