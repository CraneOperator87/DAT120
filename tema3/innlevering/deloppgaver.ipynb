c) Innledende, Bruk av funksjon og repetisjon kontrollstrukturer: Lag et script som
bruker en funksjon til å konvertere en serie med avstander i kilometer til nautiske mil.
Programmet skal inneholde en funksjon som regner fra kilometer til nautiske mil.
Programmet skal inneholde ei løkke som leser inn en avstand i kilometer, bruker
funksjonen til å konvertere den til nautiske mil, skriver ut avstanden i nautiske mil, og så
går tilbake og leser neste avstand. Scriptet skal avslutte hvis brukeren skriver inn 0.

```python
def konverter(km):
    return km / 1.852

avstanderNm = []
avstanderKm = []

print(f"For å avslutte programmet, skriv '0'.")
while True:
        try:
            kilometer = float(input("Skriv inn kilometer (Kun heltall) for å gjennomføre kilometer -> nautiske mil: "))
        except ValueError:
            print("Ugyldig input. Prøv igjen")
            continue
        if kilometer == 0:
             break

        # Legge til oppgite verdier (Kilometer)
        avstanderKm.append(kilometer)
        
        #Konvertere nautiske mil til kilometer og legge verdiene i en liste.
        nautiskMil = konverter(kilometer)
        avstanderNm.append(nautiskMil)     
        
        print(f"Oppgitte avstander i kilometer: {kilometer:.2f} Km -> {nautiskMil:.2f} Nm.")

print(f"Etter avsluttet program har du en serie med Kilometer:")
print(f"{[f'{avstand:.2f}' for avstand in avstanderKm]}")
print(f"Som blir til (Nautsike Mil):")
print(f"{[f'{avstand:.2f}' for avstand in avstanderNm]}")
```

Obligatorisk: Et perfekt tall A er et tall hvor summen av alle tallene som A er delelig med
er lik tallet selv. Inkluder 1 men ikke inkluder tallet selv. 6 er et perfekt tall fordi 6 = 1 * 2 *
3 = 1 + 2 + 3. 10 er ikke et perfekt tall siden 10 = 1 * 2 * 5, som er ulikt 1 + 2 + 5 = 8.

Del 1 – skriv en funksjon
Lag en funksjon som tar inn et tall som parameter, sjekker om tallet er perfekt og
returnerer resultatet som en boolsk verdi. Man kan finne faktorene i et tall med en for-
loop som starter på 2 og slutter på eller rett etter tallet/2 og bruk delelig-funksjonen fra
oppgave b) for å sjekke om tallet er delelig.

Del 2 – bruk funksjonen
Lag også kode som lar brukeren skrive inn et heltall og deretter bruker funksjonen din for
å sjekke om tallet er perfekt.

```python
#Fra oppgave b.
def delelig(a, b):
    return a % b == 0


def perfekt_tall(tall):
    if tall <= 1:
        return False

    sum_delere = 1

    for mulig_deler in range(2, tall // 2 + 1):
        if delelig(tall, mulig_deler):
            sum_delere += mulig_deler

    return sum_delere == tall


try:
    tall = int(input("Skriv inn et heltall for å sjekke om det er perfekt: "))

    if perfekt_tall(tall):
        print(f"Tallet {tall} er et perfekt tall.")
    else:
        print(f"Tallet {tall} er ikke et perfekt tall.")

except ValueError:
    print("Ugyldig input. Skriv inn et heltall.")
```
