Oppgave A: 
Skriv et Python-script som regner ut drivstoffprisen for en kjøretur. Du velger
selv når du skriver dette scriptet hva slags drivstoff bilen bruker. Programmet skal ta inn:
a. Pris pr. enhet drivstoff (liter bensin / liter diesel / kilowatt-time for elbil)
b. Hvor langt bilen kjører på en liter (evt. Kilowatt-time)
c. Lengde på kjøreturen

```python
# Oppgave A
type_bil = input("Hvilken type drivstoff bruker bilen din? (d/b/el): ")
if type_bil == "d":
    valgt_bil = "Dieselbil"
    pris = 23.58
    gj_forbruk = 5.5

elif type_bil == "b":
    valgt_bil = "Bensinbil"
    pris = 22.38
    gj_forbruk = 7.5

elif type_bil == "el":
    valgt_bil = "Elbil"
    pris = 4.30
    gj_forbruk = 19
else: 
    print("Ugyldig valg. Vennligst velg 'd', 'b' eller 'el'.")
    exit()

avstand = float(input("Hvor mange km skal du kjøre?: "))
if type_bil == "b" or type_bil == "d":
    forbruk = (gj_forbruk / 100) * avstand
    kostnad = forbruk * pris
    print(f"Du vil bruke {forbruk:.2f} liter drivstoff, som vil koste {kostnad:.2f} kr.")
elif type_bil == "el":
    forbruk = (gj_forbruk / 100) * avstand
    kostnad = forbruk * pris
    print(f"Du vil bruke {forbruk:.2f} kWh strøm, som vil koste {kostnad:.2f} kr.")



```
